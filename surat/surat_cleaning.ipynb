```python
import numpy as np

def convert_to_sqft(area):
    area = str(area).lower().strip()

    value = float(area.split()[0].replace(',', ''))

    if 'sqm' in area:
        return round(value * 10.7639, 2)

    elif 'sqyrd' in area:
        return round(value * 9, 2)

    elif 'sqft' in area:
        return value

    else:
        return np.nan
```

```python
bf['square_feet'] = bf['square_feet'].apply(convert_to_sqft)
```

```python
import re
def convert_price(x):

    if pd.isna(x):# if x is nan return nan
        return np.nan

    x = str(x).strip()# remove sapce

    num = re.findall(r'[\d.]+', x) # rupyee sign is not just a sign it is a Unicode character

    if not num:
        return np.nan

    num = float(num[0])

    if 'Cr' in x:
        return num * 10000000

    elif 'Lac' in x:
        return num * 100000

    else:
        return num
```

```python
bf['price'] = bf['price'].apply(convert_price)
```

```python
def clean_status(x):
    
    if pd.isna(x):
        return np.nan
        
    x = str(x)
    
    if 'Ready to Move' in x:
        return 'Ready'
        
    elif 'Resale' in x:
        return 'Resale'
        
    elif 'Poss.' in x:
        return 'Under Construction'
        
    return np.nan
```
